```python
# 1. Instalar librerías necesarias (si no las tienes)
# pyarrow es necesario para leer archivos .parquet
%pip install pandas sqlalchemy psycopg2-binary pyarrow

import pandas as pd
from sqlalchemy import create_engine

# 2. Conectar a tu base de datos (Usa las credenciales de tu docker-compose)
# engine = create_engine('postgresql://usuario:password@localhost:5432/nombre_db')
engine = create_engine('postgresql://root:root@localhost:5432/ny_taxi')

# 3. Descargar los datos (Green Taxi Nov 2025 y Zonas)
!wget https://d37ci6vzurychx.cloudfront.net/trip-data/green_tripdata_2025-11.parquet
!wget https://github.com/DataTalksClub/nyc-tlc-data/releases/download/misc/taxi_zone_lookup.csv

# 4. Cargar Zonas a Postgres
df_zones = pd.read_csv('taxi_zone_lookup.csv')
df_zones.to_sql(name='zones', con=engine, if_exists='replace')
print("Zonas cargadas exitosamente.")

# 5. Cargar Green Taxi Data a Postgres
# OJO: Aquí usamos read_parquet en lugar de read_csv
df_green = pd.read_parquet('green_tripdata_2025-11.parquet')
df_green.to_sql(name='green_taxi_trips', con=engine, if_exists='replace')
print("Viajes de Green Taxi cargados exitosamente.")
```

```python
# Ejemplo para contar viajes (adaptar según la pregunta)
query = """
SELECT count(1) 
FROM green_taxi_trips 
WHERE lpep_pickup_datetime >= '2025-11-01' 
  AND lpep_pickup_datetime < '2025-12-01'
  AND trip_distance <= 1;
"""
pd.read_sql(query, con=engine)
```
